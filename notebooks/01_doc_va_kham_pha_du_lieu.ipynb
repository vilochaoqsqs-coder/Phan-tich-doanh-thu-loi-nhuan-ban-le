```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc dữ liệu
duong_dan = "../data/raw/train.csv" if os.path.exists("../data/raw/train.csv") else "train.csv"
df = pd.read_csv(duong_dan)

# Kiểm tra thông tin cơ bản
print("Kích thước dữ liệu (dòng, cột):", df.shape)
print("\nDanh sách các cột:")
print(df.columns.tolist())
print("\nThông tin tổng quan về dữ liệu:")
print(df.info())
print("\n5 dòng dữ liệu đầu tiên:")
display(df.head())
print("\nThống kê mô tả các cột số:")
display(df.describe())
```

```python
print("Số lượng giá trị thiếu ở mỗi cột:")
print(df.isnull().sum())
print("\nTỷ lệ % giá trị thiếu:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
print("Số dòng trùng lặp:", df.duplicated().sum())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True, errors='coerce')

print("Kiểu dữ liệu sau khi chuyển đổi:")
print(df[['Order Date', 'Ship Date']].dtypes)
print("\nKiểm tra giá trị NaT (ngày không hợp lệ):")
print(df[['Order Date', 'Ship Date']].isnull().sum())
```

```python
# Vì số lượng thiếu rất ít (11 dòng), có thể xóa hoặc điền
print("Số dòng thiếu Postal Code:", df['Postal Code'].isnull().sum())

# Cách 1: Xóa các dòng thiếu (đơn giản)
df = df.dropna(subset=['Postal Code']).copy()

# Kiểm tra lại kích thước sau khi xóa
print("Kích thước sau khi xử lý:", df.shape)
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Quarter'] = df['Order Date'].dt.quarter
df['DayOfWeek'] = df['Order Date'].dt.day_name()

print("Các cột mới đã tạo:")
print(df[['Order Date', 'Year', 'Month', 'Quarter', 'DayOfWeek']].head())
```
